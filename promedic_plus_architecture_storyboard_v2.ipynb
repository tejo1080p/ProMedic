**AI-powered healthcare assistant for prescription decoding, document Q&A, and nutrition guidance.**

## Why I built it

Healthcare information is usually split across prescriptions, documents, and follow-up actions.

Promedic+ tries to connect those pieces into one workflow:

- scan a prescription
- extract structured details
- search trusted health documents
- answer with context
- generate practical nutrition guidance

## Pipeline breakdown

### 1. Prescription OCR
The app accepts an image or PDF, cleans the input, extracts text, and turns messy prescription content into structured output.

### 2. Document retrieval
Uploaded text and reference documents are chunked, embedded, and stored in a vector database so questions can be answered from context.

### 3. Context-aware response generation
The assistant combines retrieved context, patient context, and model reasoning to produce a response that feels grounded.

### 4. Nutrition guidance
This is the next workflow to build more fully: turning prescription details into simple meal timing, hydration, and interaction guidance.

## Current stack

- **Frontend:** Streamlit, Chainlit
- **LLM layer:** Ollama
- **Retrieval layer:** LangChain, MultiQueryRetriever
- **Vector store:** ChromaDB
- **OCR layer:** OpenCV, Tesseract, PDF tooling
- **Storage:** Local document folders and JSON records

## What the MVP already does

It is already usable as a small healthcare assistant: scan, search, answer, and guide.

## Upcoming feature direction

A stronger nutrition guidance workflow is the next piece I want to build so the assistant can give clearer food timing, hydration, and medication interaction suggestions.
