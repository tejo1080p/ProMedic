# Promedic+ Architecture Storyboard

A clean, standalone view of the system design and processing pipeline behind Promedic+.

This page is meant for a build-journey screenshot, not as a working app link.

```mermaid
flowchart LR
    U[User] --> UI[Streamlit / Chainlit UI]
    UI --> OCR[Prescription OCR Pipeline]
    UI --> RAG[Document RAG Assistant]
    UI --> PHARM[Pharmacy Locator]

    OCR --> PRE[Image Preprocessing]
    PRE --> TEXT[OCR Text Extraction]
    TEXT --> STRUCT[Structured Prescription Summary]
    STRUCT --> GUIDE[Nutrition / Adherence Guidance]
    STRUCT --> STORE[Local Record Storage]

    RAG --> LOAD[Document Loader]
    LOAD --> SPLIT[Chunking]
    SPLIT --> EMB[Ollama Embeddings]
    EMB --> CHROMA[Chroma Vector DB]
    CHROMA --> RETR[Multi-Query Retrieval]
    RETR --> LLM[Context-Aware Response Generation]

    PHARM --> GEO[Geolocation Capture]
    GEO --> OSM[OpenStreetMap / Overpass Lookup]
    OSM --> MAP[Map + Nearby Pharmacy Results]

    STORE --> DATA[(Local Data Folder / OCR Records)]
    CHROMA --> DB[(ChromaDB)]
```

## What the pipeline does

1. The user enters a question or uploads a prescription/document.
2. OCR cleans the prescription image or PDF and extracts readable text.
3. The extracted text is turned into a structured prescription summary.
4. Uploaded documents are chunked, embedded, and stored in a vector database.
5. Multi-query retrieval pulls the most relevant context before the LLM answers.
6. The assistant returns a grounded answer, plus follow-up guidance and source references.
7. The pharmacy module adds a practical next step by finding nearby pharmacies on a map.

## Current build stack

- Frontend: Streamlit, Chainlit
- LLM layer: Ollama
- Retrieval layer: LangChain, MultiQueryRetriever
- Vector store: ChromaDB
- OCR layer: OpenCV, Tesseract, PDF tooling
- Mapping layer: Folium, Overpass, Geopy
- Storage: Local JSON records and local document folders

## Build focus

The current MVP is designed around one workflow: scan, search, answer, and guide the user toward a practical next step.
